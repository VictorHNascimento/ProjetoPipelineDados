# Unificação dos Dados Empresa A e Empresa B

## Lendo Dados Empresa A (JSON)

```python
path_json = '../data_raw/dados_empresaA.json'
```

```python
import json
with open(path_json, 'r') as file:
    dados_empresaA = json.load(file)
```

```python
dados_empresaA[0]
```

## Lendo Dados Empresa B (CSV)

```python
path_csv = '../data_raw/dados_empresaB.csv'
```

```python
import csv
dados_empresaB = []
with open(path_csv, 'r') as file:
    spanreader =  csv.DictReader(file, delimiter=',')
    for row in spanreader:
        dados_empresaB.append(row)
```

```python
dados_empresaB[0]
```

## Padronizando Fontes

```python
empresaA_colunas = list(dados_empresaA[0].keys())
empresaB_colunas = list(dados_empresaB[0].keys())
```

```python
empresaA_colunas
```

```python
key_mapping = {
    'Nome do Item': 'Nome do Produto',
    'Classificação do Produto': 'Categoria do Produto',
    'Valor em Reais (R$)': 'Preço do Produto (R$)',
    'Quantidade em Estoque': 'Quantidade em Estoque',
    'Nome da Loja': 'Filial',
    'Data da Venda': 'Data da Venda'
}
```

```python
dados_empresaB_novos = []

for old in dados_empresaB:
    dict_temp = {}
    for key, value in old.items():
        dict_temp[key_mapping[key]] = value
    dados_empresaB_novos.append(dict_temp)
dados_empresaB_novos[0]
```

## Unificando Dados

```python
dados_unificados = []
dados_unificados.extend(dados_empresaA)
dados_unificados.extend(dados_empresaB_novos)
```

```python
nome_colunas = list(dados_unificados[-1].keys())
nome_colunas
```

```python
dados_unificados_tabela =  [nome_colunas]

for row in dados_unificados:
    l = []
    for c in nome_colunas:
        l.append(row.get(c, 'Indisponivel'))
    dados_unificados_tabela.append(l)
```

```python
#Verificando Procedimento
print(f'Quantidade de itens das empresas somadas: {len(dados_empresaB_novos+dados_empresaA)}')
print(f'Quantidade dos itens unificados: {len(dados_unificados_tabela) - 1}')
```

## Salvando Dados

```python
path_dados = '../data_processed/dados_exploracao.csv'

with open(path_dados, 'w') as file:
    writer= csv.writer(file)
    writer.writerows(dados_unificados_tabela)
```
